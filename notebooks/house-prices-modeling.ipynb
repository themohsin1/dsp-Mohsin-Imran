```python
# =========================================
# House Prices Modeling - PW1 Assignment
# =========================================

# -------------------------------
# 1. Import Libraries
# -------------------------------
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_log_error

# -------------------------------
# 2. Load Dataset
# -------------------------------
# The data files should be in the ../data/ folder
train = pd.read_csv('../data/train.csv')
test = pd.read_csv('../data/test.csv')

# Preview first 20 rows of the training data
train.head(20)

# -------------------------------
# 3. Feature Selection
# -------------------------------
# Continuous features
features_cont = ['GrLivArea', 'TotalBsmtSF']

# Categorical features
features_cat = ['Neighborhood', 'HouseStyle']

# Target variable
target = 'SalePrice'

# -------------------------------
# 4. Feature Processing
# -------------------------------
# Scale continuous features
scaler = StandardScaler()
X_train_cont = scaler.fit_transform(train[features_cont])
X_test_cont = scaler.transform(test[features_cont])

# Encode categorical features
X_train_cat = pd.get_dummies(train[features_cat], drop_first=True)
X_test_cat = pd.get_dummies(test[features_cat], drop_first=True)

# Align test columns with training columns
X_test_cat = X_test_cat.reindex(columns=X_train_cat.columns, fill_value=0)

# Combine continuous and categorical features
X_train = np.hstack([X_train_cont, X_train_cat.values])
X_test = np.hstack([X_test_cont, X_test_cat.values])

# Target
y_train = train[target].values

# -------------------------------
# 5. Model Training
# -------------------------------
model = LinearRegression()
model.fit(X_train, y_train)

# -------------------------------
# 6. Model Evaluation (RMSLE)
# -------------------------------
def compute_rmsle(y_true, y_pred, precision=2):
    """
    Compute Root Mean Squared Log Error between predicted and true values
    """
    rmsle = np.sqrt(mean_squared_log_error(y_true, y_pred))
    return round(rmsle, precision)

# Predictions on training set
y_pred_train = model.predict(X_train)
rmsle_train = compute_rmsle(y_train, y_pred_train)
print(f"Training RMSLE: {rmsle_train}")

# -------------------------------
# 7. Preview Predictions
# -------------------------------
# Display first 10 actual vs predicted values
pd.DataFrame({
    'Actual': y_train[:10],
    'Predicted': y_pred_train[:10]
})
```
